```python
#Importa la librería Numpy, Pandas, Matplotlib, OpenCV y Pillow
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cv2
from PIL import Image
```

```python
#Lea la siguiente imagen que esta en esta dirección https://imagenes2.eltiempo.com/files/image_1200_600/uploads/2019/02/22/5c705c61605df.jpeg y me muestra la imagen
import requests
from io import BytesIO

# URL de la imagen
url = "https://imagenes2.eltiempo.com/files/image_1200_600/uploads/2019/02/22/5c705c61605df.jpeg"

# Leer la imagen desde la URL
response = requests.get(url)
img = Image.open(BytesIO(response.content))

# Mostrar la imagen
plt.figure(figsize=(10, 6))
plt.imshow(img)
plt.axis('off')
plt.title('Imagen cargada desde URL')
plt.show()
```

```python
#Quiero ver el arreglo que contiene la imagen

# Convertir la imagen a un arreglo de NumPy
img_array = np.array(img)

# Ver el arreglo
print("=== ARREGLO DE LA IMAGEN ===")
print(img_array)

print("\n=== INFORMACIÓN DEL ARREGLO ===")
print(f"Forma (shape)  : {img_array.shape}")
print(f"Tipo de dato   : {img_array.dtype}")
print(f"Valor mínimo   : {img_array.min()}")
print(f"Valor máximo   : {img_array.max()}")
print(f"Total píxeles  : {img_array.size}")
```

```python
#Extrae del arreglo IMG, el arreglo que inicia en el punto 0,330 hasta el 600,830 y guárdala como img_recortada

# Recortar la imagen desde el punto (0, 330) hasta (600, 830)
# img_array[y_inicio:y_fin, x_inicio:x_fin]
img_recortada = img_array[0:600, 330:830]

# Ver información del recorte
print("=== INFORMACIÓN DE LA IMAGEN RECORTADA ===")
print(f"Forma (shape)  : {img_recortada.shape}")
print(f"Tipo de dato   : {img_recortada.dtype}")
print(f"Valor mínimo   : {img_recortada.min()}")
print(f"Valor máximo   : {img_recortada.max()}")

# Mostrar la imagen recortada
plt.figure(figsize=(8, 6))
plt.imshow(img_recortada)
plt.axis('off')
plt.title('Imagen Recortada (0,330) → (600,830)')
plt.show()
```

```python
#La imagen que ahora se llama recortada, disminuyela a un tamaño de 200 pixeles por 200 pixeles. Coloque le axis=on, imprime la imagen y el shape

# Redimensionar la imagen recortada a 200x200 píxeles
img_recortada_200 = cv2.resize(img_recortada, (200, 200))

# Imprimir el shape
print("=== INFORMACIÓN DE LA IMAGEN REDIMENSIONADA ===")
print(f"Forma (shape)  : {img_recortada_200.shape}")
print(f"Tipo de dato   : {img_recortada_200.dtype}")
print(f"Valor mínimo   : {img_recortada_200.min()}")
print(f"Valor máximo   : {img_recortada_200.max()}")

# Mostrar la imagen con axis=on
plt.figure(figsize=(5, 5))
plt.imshow(img_recortada_200)
plt.axis('on')
plt.title('Imagen Redimensionada 200x200 píxeles')
plt.xlabel('Ancho (píxeles)')
plt.ylabel('Alto (píxeles)')
plt.show()
```

```python
#Deje la imagen solo en blanco y negro, con una sola capa, guardala en una variable llamada blancoYnegro, y muestre el shape

# Convertir la imagen a escala de grises (blanco y negro, 1 sola capa)
blancoYnegro = cv2.cvtColor(img_recortada_200, cv2.COLOR_RGB2GRAY)

# Imprimir el shape
print("=== INFORMACIÓN DE LA IMAGEN EN BLANCO Y NEGRO ===")
print(f"Forma (shape)  : {blancoYnegro.shape}")
print(f"Tipo de dato   : {blancoYnegro.dtype}")
print(f"Valor mínimo   : {blancoYnegro.min()}")
print(f"Valor máximo   : {blancoYnegro.max()}")

# Mostrar la imagen en blanco y negro
plt.figure(figsize=(5, 5))
plt.imshow(blancoYnegro, cmap='gray')
plt.axis('on')
plt.title('Imagen en Blanco y Negro')
plt.xlabel('Ancho (píxeles)')
plt.ylabel('Alto (píxeles)')
plt.show()
```

```python
#Definir la ubicacion y el tamaño de la franja (en píxeles)
fila_inicio = 90
fila_fin = 105
columna_inicio = 20
columna_fin = 180

#Asignar el valor negro a esa sección para todos los canales de color
img_con_franja = img_recortada_200.copy()
img_con_franja[fila_inicio:fila_fin, columna_inicio:columna_fin] = [0, 0, 0]

#Mostrar la imagen resultante
plt.imshow(img_con_franja)
plt.title("Imagen con franja negra")
plt.axis('on')
plt.show()

print(f"Dimensiones de la imagen con franja: {img_con_franja.shape}")
```

```python
#Abrir esta imagen https://tiendalego.com.co/cdn/shop/files/77092-3_b8dae18b-a682-498a-b73a-0c89a355019b_904x904.jpg con OpenCV, no la convierta y mostra la imagen con Matplotlib 

# URL de la imagen
url2 = "https://tiendalego.com.co/cdn/shop/files/77092-3_b8dae18b-a682-498a-b73a-0c89a355019b_904x904.jpg"

# Descargar la imagen
response = requests.get(url2)
img_array = np.asarray(bytearray(response.content), dtype=np.uint8)

# Abrir con OpenCV sin convertir
img_lego = cv2.imdecode(img_array, cv2.IMREAD_UNCHANGED)

# Mostrar la imagen con Matplotlib
plt.figure(figsize=(6, 6))
plt.imshow(img_lego)
plt.axis('on')
plt.title('Imagen LEGO cargada con OpenCV')
plt.xlabel('Ancho (píxeles)')
plt.ylabel('Alto (píxeles)')
plt.show()

print("=== INFORMACIÓN DE LA IMAGEN ===")
print(f"Forma (shape)  : {img_lego.shape}")
print(f"Tipo de dato   : {img_lego.dtype}")
print(f"Valor mínimo   : {img_lego.min()}")
print(f"Valor máximo   : {img_lego.max()}")
```

```python
#Lea nuevamente la imagen de la URL1 con OpenCV, y utiliza el método que convierte BGR a RGB y muestre la imagen. Igualmente, muestrame la forma de la imagen

# Leer con OpenCV
img_bgr = cv2.imdecode(img_array, cv2.IMREAD_UNCHANGED)

# Convertir de BGR a RGB
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Mostrar la imagen
plt.figure(figsize=(10, 6))
plt.imshow(img_rgb)
plt.axis('on')
plt.title('Imagen URL1 - BGR convertida a RGB')
plt.xlabel('Ancho (píxeles)')
plt.ylabel('Alto (píxeles)')
plt.show()

# Mostrar la forma
print("=== INFORMACIÓN DE LA IMAGEN ===")
print(f"Forma BGR  (shape) : {img_bgr.shape}")
print(f"Forma RGB  (shape) : {img_rgb.shape}")
print(f"Tipo de dato       : {img_rgb.dtype}")
print(f"Valor mínimo       : {img_rgb.min()}")
print(f"Valor máximo       : {img_rgb.max()}")
```

```python
#Toma la imagen anterior RGB y conviertela en escala de grises. Muestrame el shape e imprímela

# Convertir la imagen RGB a escala de grises
img_gris = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

# Mostrar el shape
print("=== INFORMACIÓN DE LA IMAGEN EN ESCALA DE GRISES ===")
print(f"Shape original RGB : {img_rgb.shape}")
print(f"Shape escala grises: {img_gris.shape}")
print(f"Tipo de dato       : {img_gris.dtype}")
print(f"Valor mínimo       : {img_gris.min()}")
print(f"Valor máximo       : {img_gris.max()}")

# Mostrar la imagen en escala de grises
plt.figure(figsize=(10, 6))
plt.imshow(img_gris, cmap='gray')
plt.axis('on')
plt.title('Imagen en Escala de Grises')
plt.xlabel('Ancho (píxeles)')
plt.ylabel('Alto (píxeles)')
plt.colorbar(label='Intensidad (0=Negro, 255=Blanco')
plt.show()
```

```python
#De imagen_rgb mantenga la capa roja, y la matriz de la capa verde y la capa azul llénalas con 0, y muestra el resultado

# Copiar la imagen RGB para no modificar la original
img_solo_rojo = img_rgb.copy()

# Llenar la capa Verde (canal 1) con 0
img_solo_rojo[:, :, 1] = 0

# Llenar la capa Azul (canal 2) con 0
img_solo_rojo[:, :, 2] = 0

# Mostrar el resultado
plt.figure(figsize=(10, 6))
plt.imshow(img_solo_rojo)
plt.axis('on')
plt.title('Imagen - Solo Canal Rojo (G=0, B=0)')
plt.xlabel('Ancho (píxeles)')
plt.ylabel('Alto (píxeles)')
plt.show()

print("=== INFORMACIÓN DE LA IMAGEN ===")
print(f"Shape             : {img_solo_rojo.shape}")
print(f"Canal Rojo   [0]  : min={img_solo_rojo[:,:,0].min()}, max={img_solo_rojo[:,:,0].max()}")
print(f"Canal Verde  [1]  : min={img_solo_rojo[:,:,1].min()}, max={img_solo_rojo[:,:,1].max()}")
print(f"Canal Azul   [2]  : min={img_solo_rojo[:,:,2].min()}, max={img_solo_rojo[:,:,2].max()}")
```

# Observación
**Una convolución es tomar una matriz y multiplicarla por una más grande**

**Un kernel es una matriz pequeña que generalmente es impar, solo puede tomar valores entre -6 y 6**

**Cuando se hace una convolución con Kernel(3,3), la matriz pierde 2 filas y 2 columnas**

**Cuando se hace con Kernel(3,3) con Padding, la matriz mantiene el mismo número de columnas y filas, pero con 2 de cada una con valores en 0**

**Cuando se hace una convolución con Kernel(5,5), la matriz pierde 4 filas y 4 columnas**

```python
#Cargar la siguiente imagen de la siguiente dirección https://fotografias-neox.atresmedia.com/clipping/cmsimages02/2020/11/09/FDAD0B9D-4486-4CE7-9C2B-6CC195BA1977/98.jpg y cárgala en RGB, muestrame la imagen y muestrame el shape

link = 'https://fotografias-neox.atresmedia.com/clipping/cmsimages02/2020/11/09/FDAD0B9D-4486-4CE7-9C2B-6CC195BA1977/98.jpg'

resp = requests.get(link)
img_3 = Image.open(BytesIO(response.content)).convert('RGB')

img_arr = np.array(img_3)

print('Shape de la imagen (alto, ancho, canales):', img_arr.shape)

plt.imshow(img_arr)
plt.axis('off')
plt.title('Imagen cargada en RGB')
plt.show()
```

```python
#Cree una matriz con los siguientes valores [[2,1,-2,0,1],[3,-3,2,5,4],[-2,0,1,2,2],[4,-5,3,2,0],[1,4,3,0,1]] y llamalo ardila

# Crear la matriz ardila
ardila = [[2, 1, -2, 0, 1],
          [3, -3, 2, 5, 4],
          [-2, 0, 1, 2, 2],
          [4, -5, 3, 2, 0],
          [1, 4, 3, 0, 1]]

# Mostrar la matriz
print("Matriz ardila:")
for fila in ardila:
    print(fila)
    
print(f"\nShape de la matriz: {np.array(ardila).shape}")
```

```python
#Utilice el kernel ardila y haga una operacion convuncional usando la librería de OpenCV llamada Filter2D, guardela en una variable y muestre el shape del resultado

# Aplicar la operación convolucional con filter2D
# Nota: La imagen tiene 3 canales (RGB), filter2D aplicará el kernel a cada canal
imagen_conv = cv2.filter2D(img_arr, -1, np.array(ardila))

# Visualizar los resultados (opcional)
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 2)
plt.imshow(imagen_conv)
plt.title('Imagen después de convolución con kernel ardila')
plt.axis('off')
plt.tight_layout()
plt.show()
```
